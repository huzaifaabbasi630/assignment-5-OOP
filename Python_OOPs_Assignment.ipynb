**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2
class BankAccount:
    def __init__(self , initial_balance = 0):
        self.__balance = initial_balance
        
    def deposit(self , amount):
        if amount > 0:
             self.__balance += amount
             print("successflly deposit")
        else:
            print("invalid deposit")
    
    def withdraw(self , amount):
        if amount <= 0 :
            print("invalid balance")
        elif amount > self.__balance:
            print(f'transaction failed Your current balance is {self.__balance}')
        else:
            self.__balance -= amount
            print('withdraw success')
    def gett_balance(self):
        return self.__balance

Bnk_Acnt1 = BankAccount(1000)
print(f"Initial Balance: {Bnk_Acnt1.gett_balance()}")
Bnk_Acnt1.deposit(500)
print(f'after deposit you balance is :{Bnk_Acnt1.gett_balance()}')
Bnk_Acnt1.withdraw(300)
print(f'after withdraw your deposit is :{Bnk_Acnt1.gett_balance()}')
Bnk_Acnt1.withdraw(1500)
print(f"Final Balance: {Bnk_Acnt1.gett_balance()}")
```

**Q3:** Aik class banayein `Employee` jismein salary private ho. Agar koi salary ko minus (negative) value set karne ki koshish kare, toh setter method 'Invalid Salary' ka error message show kare.

```python
# Write your code here for Q3
class Employee:
    def __init__(self, salary=0):
        self.__salary = 0
        self.set_salary(salary)
        
    def set_salary(self, salary):
        if salary < 0:
            print("Invalid Salary")
        else:
            self.__salary = salary
            
    def get_salary(self):
        return self.__salary

# Test
emp = Employee(50000)
print(f"Salary: {emp.get_salary()}")

emp.set_salary(-5000)  # Yeh error show karega
print(f"Salary after invalid update: {emp.get_salary()}")
```

**Q7:** Multilevel Inheritance ka use karein. Class `Animal` -> Class `Mammal` -> Class `Dog`. Dog class ke object se parent classes ke features ko access karein.

```python
# Write your code here for Q7
class Animal:
    def eat(self):
        return "Eating..."

class Mammal(Animal):
    def walk(self):
        return "Walking..."

class Dog(Mammal):
    def bark(self):
        return "Barking..."

dog = Dog()
print(dog.eat())   # Parent
print(dog.walk())  # Parent's parent
print(dog.bark())  # Child
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"

p1 = Point(2, 3)
p2 = Point(4, 5)
print("Sum of points:", p1 + p2)
```

**Q16:** `abc` module ka use karte hue aik abstract class `Computer` banayein jisme abstract method `process()` ho. Is class ka object banane ki koshish karein aur dekhein kya hota hai.

```python
# Write your code here for Q16
from abc import ABC, abstractmethod

class Computer(ABC):
    @abstractmethod
    def process(self):
        pass

# Object banane ki koshish (TypeError dega)
try:
    c = Computer()
except TypeError as e:
    print("Error:", e)
```

**Q17:** Q16 wali `Computer` class ko inherit karke do child classes `Laptop` aur `Desktop` banayein aur dono mein `process()` method ko mandatory implement karein.

```python
# Write your code here for Q17
from abc import ABC, abstractmethod

class Computer(ABC):
    @abstractmethod
    def process(self):
        pass

class Laptop(Computer):
    def process(self):
        print("Laptop processing data using Battery.")

class Desktop(Computer):
    def process(self):
        print("Desktop processing data using Direct Power.")

Laptop().process()
Desktop().process()
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
# Write your code here for Q18
from abc import ABC, abstractmethod

class Car(ABC):
    def start_engine(self):  # Concrete method
        print("Engine started... Vroom!")

    @abstractmethod
    def drive(self):  # Abstract method
        pass

class SportsCar(Car):
    def drive(self):
        print("Driving sports car at high speed!")

car = SportsCar()
car.start_engine()
car.drive()
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def make_payment(self, amount):
        pass

class CreditCardPayment(Payment):
    def make_payment(self, amount):
        print(f"Paid PKR {amount} securely using Credit Card.")

class UPIPayment(Payment):
    def make_payment(self, amount):
        print(f"Paid PKR {amount} instantly using UPI / Mobile Wallet.")

# Testing payment system
payments = [CreditCardPayment(), UPIPayment()]
for p in payments:
    p.make_payment(1500)
```
